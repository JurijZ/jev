```python
import requests
import json
import os
```

```python
# Load environment variables from .env file
from dotenv import load_dotenv
load_dotenv()
```

```python
# import the OPENROUTER_API_KEY from the .env file
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
```

```python


# The model answers narrow, typed questions about the state. Your code owns the workflow.
response = requests.post(
  url="https://openrouter.ai/api/alpha/decisions",
  headers={
    "Authorization": f"Bearer {OPENROUTER_API_KEY}",
    "Content-Type": "application/json",
  },
  data=json.dumps({
    "model": "microsoft/microsoft-decision-1",
    "state": "Mano mokėjimai neveikia jau kelias dienas. Kuo skubiau sutvarkykite! Kiek galima laukti?",
    "questions": {
      "is_urgent": {
        "type": "noul",
        "instructions": "Ar žnutė reikalauja skubumo?",
        "criteria": {
          "true": "Explicitly time-sensitive",
          "false": "No urgency expressed"
        }
      },
      "department": {
        "type": "choice",
        "instructions": "Kurį komandą reikia priskirti šį uždavinį?",
        "criteria": {
          "billing": "Mokėjimai, sąskaitos, grąžinimai",
          "technical": "Klaidos, išjungimai, integracijos",
          "sales": "Kainodara, nauji vartotojai, pasiūlymai",
          "other": "Kita"
        }
      },
      "frustration": {
        "type": "score",
        "instructions": "Ant kiek susierzinęs yra klientas?",
        "criteria": ["Ramus", "Nepatenkintas", "Piktas"]
      }
    }
  })
)
```

```python
answers = response.json()["answers"]
# noul is a probability from 0 (no) to 1 (yes); choice and score carry the full distribution.
print(answers["is_urgent"]["noul"])
print(answers["department"]["choice"], answers["department"]["probabilities"])
print(answers["frustration"]["score"])
```
